````python
"""Q99.**  
Implement **double normalization** (min-max then Z-score) on an ill-scaled dataset:
```python
X = np.random.rand(50, 4) * np.array([1, 100, 1000, 0.01])
```
Step 1: Min-max scale each column → all values in `[0, 1]`  
Step 2: Z-score standardize the min-max scaled data  

After Step 2:
- Are all column means exactly 0? (Yes)
- Are all column stds exactly 1? (Yes, if std > 0)
- What would happen if a column had all identical values (std = 0)? Handle this case safely.""" 

X = np.random.rand(50, 4) * np.array([1, 100, 1000, 0.01])

# Inject a zero-variance column to test protection logic
X[:, 0] = 5.0 

# Step 1: Min-Max Normalization
col_min = X.min(axis=0)
col_max = X.max(axis=0)
range_diff = col_max - col_min

# Fallback pattern if max == min to avoid division-by-zero errors
range_diff_safe = np.where(range_diff == 0, 1.0, range_diff)
X_minmax = (X - col_min) / range_diff_safe

# Step 2: Z-Score Standardization
means = X_minmax.mean(axis=0)
stds = X_minmax.std(axis=0)

# If standard deviation is zero, bypass scaling division
stds_safe = np.where(stds == 0, 1.0, stds)
X_zscore = (X_minmax - means) / stds_safe

print("Standardized Output Array (First 3 Rows):\n", X_zscore[:3])
````

````python
"""Q100.**  
**Full ML pipeline challenge** — implement everything from scratch using only Sections 1–13:

```python
X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)  # binary target
```

1. **Z-score standardize** `X` column-wise using broadcasting
2. **Shuffle and split** 70/15/15 using `np.argsort(np.random.rand(200))` as shuffle index
3. **Initialize weights** `w = np.zeros(5)` and bias `b = 0.0`
4. **Forward pass** on `X_train`: `z = X_train @ w + b`, `y_hat = 1 / (1 + np.exp(-z))`
5. **BCE loss**: clip `y_hat` to `[1e-7, 1-1e-7]`, compute `-np.mean(y*log + (1-y)*log)`
6. **Gradient**: `grad_w = X_train.T @ (y_hat - y_train) / len(y_train)`
7. **One gradient descent step**: `w -= 0.1 * grad_w`
8. **Forward pass again** with updated `w`, compute new BCE loss
9. Print: initial loss, updated loss (should be slightly lower), gradient norm `np.sqrt((grad_w**2).sum())` """ 

# Setup seed for reproducible evaluations
np.random.seed(42)

X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(float) # Binary assignment target matching training operations

# 1. Standardization via Broadcasting
X_scaled = (X - X.mean(axis=0)) / X.std(axis=0)

# 2. Shuffle Split Index Generation
shuffle_idx = np.argsort(np.random.rand(200))
X_shuffled = X_scaled[shuffle_idx]
y_shuffled = y[shuffle_idx]

X_train, y_train = X_shuffled[:140], y_shuffled[:140]

# 3. Model Weight Initialization
w = np.zeros(5)
b = 0.0

# 4. Forward Propagation Pass 1
z1 = X_train @ w + b
y_hat1 = 1 / (1 + np.exp(-z1))

# 5. Calculate Evaluation Losses
eps = 1e-7
y_hat1_clipped = np.clip(y_hat1, eps, 1 - eps)
initial_loss = -np.mean(y_train * np.log(y_hat1_clipped) + (1 - y_train) * np.log(1 - y_hat1_clipped))

# 6. Gradient Backpropagation calculation
grad_w = (X_train.T @ (y_hat1 - y_train)) / len(y_train)

# 7. Optimizer Updates
w -= 0.1 * grad_w

# 8. Forward Propagation Pass 2
z2 = X_train @ w + b
y_hat2 = np.clip(1 / (1 + np.exp(-z2)), eps, 1 - eps)
updated_loss = -np.mean(y_train * np.log(y_hat2) + (1 - y_train) * np.log(1 - y_hat2))

# 9. Pipeline Evaluation Metric Summary Output
grad_norm = np.sqrt(np.sum(grad_w**2))
print(f"Initial BCE Loss value:      {initial_loss:.6f}")
print(f"Updated BCE Loss value:      {updated_loss:.6f} (Lower: {updated_loss < initial_loss})")
print(f"L2 Norm weight matrix gradient: {grad_norm:.6f}")
````
