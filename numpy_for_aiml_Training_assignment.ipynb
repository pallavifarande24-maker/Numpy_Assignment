````python
"""*Q1.** Given model confidence scores:  
`predictions = [0.72, 0.31, 0.95, 0.58, 0.84, 0.12, 0.67, 0.90]`  
- Sort in **descending order** (highest confidence first).  
- Use `np.argsort` to get the **top-3 prediction indices**.  
- Assign ranks (1 = most confident) to each prediction."""

import numpy as np 
predictions = np.array([0.72, 0.31, 0.95, 0.58, 0.84, 0.12, 0.67, 0.90])
sorted_preds = np.sort(predictions)[::-1] 
top_3_indices = np.argsort(predictions)[::-1][:3] 
ranks = len(predictions) - np.argsort(np.argsort(predictions))

print(f"Sorted (Descending): {sorted_preds}")
print(f"Top-3 Indices:       {top_3_indices}")
print(f"Ranks:               {ranks}")

````

````python
""" *Q1.** You have a large array of 2 million values. Compare the time to:  
- Calculate `np.sin(arr) + np.cos(arr)` using a Python `for` loop  
- Calculate it vectorized  
Report the speedup ratio.""" 

import numpy as np 
import time
arr = np.random.rand(2_000_000)
start_loop = time.time()
loop_result = np.zeros_like(arr)
for i in range(len(arr)):
    loop_result[i] = np.sin(arr[i]) + np.cos(arr[i])
end_loop = time.time()
loop_time = end_loop - start_loop

# 2. Vectorized implementation
start_vec = time.time()
vec_result = np.sin(arr) + np.cos(arr)
end_vec = time.time()
vec_time = end_vec - start_vec

# 3. Report speedup ratio
speedup = loop_time / vec_time
print(f"Loop Time:       {loop_time:.4f} seconds")
print(f"Vectorized Time: {vec_time:.4f} seconds")
print(f"Speedup Ratio:   {speedup:.1f}x faster")
````

````python
"""Q2.** Simulate a **train/validation/test split** (70/15/15) on 500 samples:
- Shuffle indices with `rng.permutation`.
- Slice into three groups.
- Confirm no indices overlap between splits.""" 

rng = np.random.default_rng(7)
num_samples = 500

# 1. Shuffle indices
indices = np.arange(num_samples)
shuffled_indices = rng.permutation(indices)

# 2. Slice into 70/15/15 groups
# 70% of 500 = 350
# 15% of 500 = 75 (350 + 75 = 425)
train_idx = shuffled_indices[:350]
val_idx = shuffled_indices[350:425]
test_idx = shuffled_indices[425:]

# 3. Confirm no indices overlap using sets
overlap_check = (
    len(set(train_idx).intersection(set(val_idx))) == 0 and
    len(set(val_idx).intersection(set(test_idx))) == 0 and
    len(set(train_idx).intersection(set(test_idx))) == 0
)
print(f"Train size: {len(train_idx)}, Val size: {len(val_idx)}, Test size: {len(test_idx)}")
print(f"No overlapping indices? {overlap_check}") 
````

````python
""" Q3.** Implement a **Monte Carlo estimate of π**:
- Generate N=1,000,000 random (x, y) points in the unit square using `rng.random`.
- Count how many fall inside the unit circle: $x^2 + y^2 \leq 1$.
- Estimate π as `4 * inside / N`. How close is it to `np.pi`?"""  

rng = np.random.default_rng(7)
N = 1_000_000

# 1. Generate random (x, y) points in the unit square [0, 1)
# Generating a shape of (N, 2) creates both x and y coordinates at once
points = rng.random((N, 2))
x = points[:, 0]
y = points[:, 1]

# 2. Count how many fall inside the unit circle
inside_circle = (x**2 + y**2) <= 1
num_inside = np.sum(inside_circle)

# 3. Estimate pi
pi_estimate = 4 * num_inside / N
difference = abs(pi_estimate - np.pi)
print(f"Estimated π: {pi_estimate}")
print(f"Actual np.pi: {np.pi}")
print(f"Absolute Error: {difference:.6f}")
````

````python
""" Q1.** Solve the following linear system using `np.linalg.solve`:

$$3x + 2y - z = 7$$
$$x - y + 4z = 3$$
$$2x + 5y - 2z = 1$$

Verify by substituting the result back into the equations.""" 

import numpy as np
A = np.array([[3, 2, -1], [1, -1, 4], [2, 5, -2]], dtype=float)
b = np.array([7, 3, 1], dtype=float)
# 2. Solve for [x, y, z]
xyz = np.linalg.solve(A, b)

# 3. Verify by substituting back (Ax should equal b)
verification = np.allclose(np.dot(A, xyz), b)
print(f"Solution: x = {xyz[0]:.4f}, y = {xyz[1]:.4f}, z = {xyz[2]:.4f}")
print(f"Verification successful? {verification}")
````

````python
""" Q2.** Generate a `(50, 4)` dataset where:  
- Features are random normal data  
- True relationship: `y = 1.5*x1 - 2.0*x2 + 0.5*x3 + 3.0*x4 + noise`  

Estimate weights using the Normal Equation with a bias term. Report MSE.""" 

rng = np.random.default_rng(10)
X_raw = rng.standard_normal((50, 4))
true_weights = np.array([1.5, -2.0, 0.5, 3.0])
noise = rng.normal(scale=0.1, size=50) 
y = X_raw @ true_weights + noise

# 3. Add a bias term (a column of ones) to the feature matrix
X_bias = np.c_[np.ones(50), X_raw]

# 4. Apply the Normal Equation: theta = inverse(X_bias.T @ X_bias) @ X_bias.T @ y
theta = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y

# 5. Compute Mean Squared Error (MSE)
y_pred = X_bias @ theta
mse = np.mean((y - y_pred)**2)

print("True weights:  [bias=0.0, w1=1.5, w2=-2.0, w3=0.5, w4=3.0]")
print(f"Estimated $\\theta$: {theta}")
print(f"Mean Squared Error (MSE): {mse:.6f}") 
````

````python
"""**Q3.** Compute cosine similarity between all pairs of these three word vectors (simplified word embeddings):
```python
king   = np.array([0.8, 0.3, 0.5])
queen  = np.array([0.7, 0.4, 0.6])
apple  = np.array([0.1, 0.9, 0.2])
```
Which pair is most similar? Does this match your intuition?"""  

king  = np.array([0.8, 0.3, 0.5])
queen = np.array([0.7, 0.4, 0.6])
apple = np.array([0.1, 0.9, 0.2])
def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

sim_king_queen = cosine_similarity(king, queen)
sim_king_apple = cosine_similarity(king, apple)
sim_queen_apple = cosine_similarity(queen, apple)

print(f"Similarity (King, Queen): {sim_king_queen:.4f}")
print(f"Similarity (King, Apple): {sim_king_apple:.4f}")
print(f"Similarity (Queen, Apple): {sim_queen_apple:.4f}")
````

````python
""" *Q1.** You receive the following housing dataset (size_m2, num_rooms, price_k€):
```python
houses = np.array([
    [45,  1,  120], [90,  3,  250], [65,  2,  175],
    [120, 4,  380], [30,  1,   80], [200, 5,  600],
])
```
- Apply **min-max scaling** to all three features.
- Apply **Z-score standardization** to all three features.
- Which scaling would you use to feed into a neural network, and why?""" 


import numpy as np
houses = np.array([
    [45,  1,  120], [90,  3,  250], [65,  2,  175],
    [120, 4,  380], [30,  1,   80], [200, 5,  600],
], dtype=float)
# 1. Min-Max Scaling: (X - min) / (max - min)
min_vals = houses.min(axis=0)
max_vals = houses.max(axis=0)
min_max_scaled = (houses - min_vals) / (max_vals - min_vals)

# 2. Z-score Standardization: (X - mean) / std
mean_vals = houses.mean(axis=0)
std_vals = houses.std(axis=0)
z_score_scaled = (houses - mean_vals) / std_vals

print("--- Min-Max Scaled Dataset (Bounded 0-1) ---")
print(np.round(min_max_scaled, 3))
print("\n--- Z-score Standardized Dataset (Mean=0, Std=1) ---")
print(np.round(z_score_scaled, 3)) 
````

````python
""" Q3.** Build a complete preprocessing pipeline:
- Create a `(300, 5)` feature matrix with features at very different scales  
  (e.g., multiply each column by `[1, 10, 100, 0.1, 1000]`)
- Standardize all features
- Verify that each column has mean ≈ 0 and std ≈ 1
- Split into 60% train, 20% val, 20% test — confirm no overlap """  

rng = np.random.default_rng(42)
X_raw = rng.standard_normal((300, 5)) * np.array([1, 10, 100, 0.1, 1000])
# your code here... 

print("Original Means:", np.round(X_raw.mean(axis=0), 2))
print("Original Stds: ", np.round(X_raw.std(axis=0), 2))

# 2. Data Split (60% Train, 20% Val, 20% Test)
# Data leakage se bachne ke liye standardisation se PEHLE split karna chahiye!
N = X_raw.shape[0]
shuffled_indices = rng.permutation(N)

train_cutoff = int(0.60 * N)        # 180
val_cutoff = int((0.60 + 0.20 * N)) # 180 + 60 = 240

train_idx = shuffled_indices[:train_cutoff]
val_idx   = shuffled_indices[train_cutoff:val_cutoff]
test_idx  = shuffled_indices[val_cutoff:]

X_train_raw = X_raw[train_idx]
X_val_raw   = X_raw[val_idx]
X_test_raw  = X_raw[test_idx]

# 3. Standardization (Z-score)
# Rule: Mean aur Std sirf Train set se nikalenge takis validation/test leak na ho
X_train_mean = X_train_raw.mean(axis=0)
X_train_std  = X_train_raw.std(axis=0)

# Formula apply karenge teeno splits par
X_train_scaled = (X_train_raw - X_train_mean) / X_train_std
X_val_scaled   = (X_val_raw - X_train_mean) / X_train_std
X_test_scaled  = (X_test_raw - X_train_mean) / X_train_std

print("\n--- Step 4: Verification ---")
print("Train Scaled Means (Target 0):", np.abs(np.round(X_train_scaled.mean(axis=0), 5)))
print("Train Scaled Stds  (Target 1):", np.round(X_train_scaled.std(axis=0), 5))

# 5. Overlap Confirmation Check
has_overlap = (
    len(set(train_idx) & set(val_idx)) > 0 or
    len(set(val_idx) & set(test_idx)) > 0 or
    len(set(train_idx) & set(test_idx)) > 0
)

print(f"\nAny index overlap? {has_overlap}")
print(f"Train Shape: {X_train_scaled.shape}, Val Shape: {X_val_scaled.shape}, Test Shape: {X_test_scaled.shape}")

````
