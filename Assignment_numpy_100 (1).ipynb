````python
#29. How to round away from zero a float array ? (★☆☆)
#`hint: np.uniform, np.copysign, np.ceil, np.abs, np.where` 

import numpy as np

# Create a sample float array with positive and negative values
Z = np.array([2.1, -2.1, 1.5, -1.5, 0.0, 4.9, -4.9])

# Step 1: Get the absolute values
abs_Z = np.abs(Z)

# Step 2: Round up towards positive infinity
ceil_Z = np.ceil(abs_Z)

# Step 3: Copy the original signs back to the rounded numbers
rounded_Z = np.copysign(ceil_Z, Z)

print("Original Array: ", Z)
print("Rounded Away:   ", rounded_Z)
````

````python
## 63. Create an array class that has a name attribute (★★☆)`hint: class method`

import numpy as np


class NamedArray(np.ndarray):

    def __new__(cls, array, name="pallavi"):
        obj = np.asarray(array).view(cls)
        obj.name = name
        return obj

    def __array_finalize__(self, obj):
        if obj is None:
            return
        self.name = getattr(obj, "name", "pallavi")


A = NamedArray(np.arange(10), name="farande")
print(A)
print(A.name)
````

````python
# 79. Consider 2 sets of points P0,P1 describing lines (2d) and a set of points P, how to compute distance from each point j (P[j]) to each line i (P0[i],P1[i])? (★★★)
#`No hints provided...` 

import numpy as np

P0 = np.random.uniform(-10, 10, (10, 2))
P1 = np.random.uniform(-10, 10, (10, 2))
P = np.random.uniform(-10, 10, (5, 2))

L = P1 - P0
len_sq = (L**2).sum(axis=1)
dist = np.zeros((len(P), len(P0)))
for i, p in enumerate(P):
    u = ((p - P0) * L).sum(axis=1) / len_sq
    u = np.clip(u, 0, 1)
    proj = P0 + u[:, None] * L
    dist[i] = np.sqrt(((proj - p) ** 2).sum(axis=1))
print(dist)
````

````python
#### 85. Create a 2D array subclass such that Z[i,j] == Z[j,i] (★★★)
#`hint: class method` 

import numpy as np


class SymetricArray(np.ndarray):

    def __setitem__(self, index, value):
        i, j = index
        super().__setitem__((i, j), value)
        super().__setitem__((j, i), value)


def symetric(Z):
    return np.asarray(Z + Z.T - np.diag(Z.diagonal())).view(SymetricArray)



base_matrix = np.zeros((3, 3), dtype=int)

S = symetric(base_matrix)

print("base_matrix:")
print(S)

S[0, 1] = 5

print("\n After change the value:")
print(S)
````

````python
# 87. Consider a 16x16 array, how to get the block-sum (block size is 4x4)? (★★★)
#`hint: np.add.reduceat, from numpy.lib.stride_tricks import sliding_window_view (np>=1.20.0)

import numpy as np

# Create a 16x16 array of ones
a = np.ones((16, 16))
b= 4

# Method 1: Using reshape and sum
S = a.reshape(a.shape[0] // b, b, a.shape[1] // b, b).sum(axis=(1, 3))
print(S)

# Method 2: Using np.add.reduceat
# row_indices = [0, 4, 8, 12]
row_indices = np.arange(0, a.shape[0], b)
# col_indices = [0, 4, 8, 12]
col_indices = np.arange(0, a.shape[1], b)

res = np.add.reduceat(
    np.add.reduceat(a, row_indices, axis=0), col_indices, axis=1)

print(res)
````

````python
# 88. How to implement the Game of Life using numpy arrays? (★★★)
#`No hints provided...` 

import numpy as np


def iterate(Z):
    N = (
        Z[0:-2, 0:-2]
        + Z[0:-2, 1:-1]
        + Z[0:-2, 2:]
        + Z[1:-1, 0:-2]
        + Z[1:-1, 2:]
        + Z[2:, 0:-2]
        + Z[2:, 1:-1]
        + Z[2:, 2:]
    )
    birth = (N == 3) & (Z[1:-1, 1:-1] == 0)
    survive = ((N == 2) | (N == 3)) & (Z[1:-1, 1:-1] == 1)
    Z[...] = 0
    Z[1:-1, 1:-1][birth | survive] = 1
    return Z


Z = np.random.randint(0, 2, (50, 50))
for i in range(100):
    Z = iterate(Z)
print(Z)
````

````python
#### 90. Given an arbitrary number of vectors, build the cartesian product (every combination of every item) (★★★)
#`hint: np.indices` 

import numpy as np


def cartesian(arrays):
    arrays = [np.asarray(a) for a in arrays]
    shape = (len(x) for x in arrays)
    ix = np.indices(shape, dtype=int)
    ix = ix.reshape(len(arrays), -1).T
    for n, arr in enumerate(arrays):
        ix[:, n] = arrays[n][ix[:, n]]
    return ix


print(cartesian(([1, 2, 3], [4, 5], [6, 7])))
````

````python
#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
#`hint: np.cumsum, np.interp 

import numpy as np

X = np.array([0, 1, 2, 4, 7, 10])
Y = np.array([0, 2, 3, 3, 5, 8])
dims = np.column_stack((X, Y))
segments = np.diff(dims, axis=0)
d_segments = np.sqrt((segments**2).sum(axis=1))
cumulative_d = np.concatenate(([0], np.cumsum(d_segments)))
total_d = cumulative_d[-1]
n_samples = 10
equidistant_d = np.linspace(0, total_d, n_samples)
X_samples = np.interp(equidistant_d, cumulative_d, X)
Y_samples = np.interp(equidistant_d, cumulative_d, Y) 
print(equidistant_d)
````
